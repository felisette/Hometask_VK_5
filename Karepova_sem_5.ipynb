### Задание 1. 
Восстановление бинарного дерева из массива 

Объясните, как определяется связь родитель–потомок и какая структура данных используется для 
хранения узлов.
Ответ: для хранения узлов используется класс, поля которого включают в себя указатели на правого и левого потомка и значение. Предполагается, что входной массив в ячейке с нулевым индексом собержит корень, правый потомок для родителя в ячейке с индексом i хранится в ячейке с индексом 2i+2, левый - 2i+1.

```python
class TreeNode:
    def __init__(self, val = 0, left = None, right = None):
        self.data = val
        self.left = left
        self.right = right
```

```python
def buildTree(arr, i):
    if i>=len(arr):
        return None
    root = TreeNode(arr[i])
    root.left = buildTree(arr, 2*i + 1)
    root.right = buildTree(arr, 2*i + 2)
    return root

def printTree(node):
    if node:
        printTree(node.left)
        print(node.data, end=" ")
        printTree(node.right)
```

### Задание 2 
Симметричное бинарное дерево: BFS DFS

Поясните, какие пары узлов сравниваются и почему алгоритм работает 
корректно. Ответ: Сравниваются пары queue[i] и queue[queueLen - i - 1] — то есть элементы текущего уровня с противоположных концов. Алгоритм корректен, потому что на каждом уровне он проверяет, что дерево — «зеркало» самого себя, и добавляет детей в правильном порядке для проверки следующего уровня.

```python
def isSymmetricBFS(root):
    if root==None:
        return True
    queue = [root]
    while queue:
        queueLen = len(queue)
        for i in range(0, queueLen):
            if queue[i]==None and queue[queueLen - i - 1]==None:
                continue
            if queue[i]==None or queue[queueLen - i - 1]==None:
                return False
            if queue[i].data!=queue[queueLen - i - 1].data:
                return False
            queue.append(queue[i].left)
            queue.append(queue[i].right)
        queue = queue[queueLen:]
    return True

def deptSearch(root, res):
    if root==None:
        return res
    deptSearch(root.left, res)
    res.append(root.data)
    deptSearch(root.right, res)
    return res

def isSymmetricDFS(root):
    if root==None:
        return True
    data = []
    data = deptSearch(root, data)
    j = len(data) - 1
    for i in range(0, len(data)//2):
        if data[i]!=data[j]:
            return False
        j -= 1
    return True
```

```python
val = [9, 8, 8, 3, 6, 6, 3]
root = buildTree(val, 0)
print("BFS", isSymmetricBFS(root), "\nDFS", isSymmetricDFS(root))

val = [3, 7, 8, 9, 6, 9]
root = buildTree(val, 0)
print("BFS", isSymmetricBFS(root), "\nDFS", isSymmetricDFS(root))
```

### Задание 3
Поиск минимальной глубины бинарного дерева

Объясните, почему поиск в ширину (BFS) позволяет корректно найти минимальную глубину. Ответ: BFS находит минимальную глубину правильно, потому что обходит дерево по уровням: как только он впервые встречает лист, это лист с наименьшей глубиной, так как остальные уровни не имели листов, значит имели потомков, значит глубина для них как минимум на 1 больше.

```python
def minDepth(root):
    if root == None:
        return 0
    if root.left == None and root.right == None:
        return 1
    if root.left != None and root.right != None:
        return 1 + min(minDepth(root.left), minDepth(root.right))
    if root.left != None:
        return 1 + minDepth(root.left)
    if root.right != None:
        return 1 + minDepth(root.right)
```

```python
val = [9, 8, 8, 3, 6, 6, 3]
root = buildTree(val, 0)
print("Минимальная глубина ", minDepth(root))

val = [3, 7, 8, 9, 6, 9]
root = buildTree(val, 0)
print("Минимальная глубина ", minDepth(root))

val = [11, 6, 16, 2, 9]
root = buildTree(val, 0)
print("Минимальная глубина ", minDepth(root))
```

### Задание 4
Поиск произведения максимального и минимального элементов

Поясните, почему поиск крайних элементов выполняется за O(h), где h — высота дерева. Ответ: дерево поиска строится таким образом, что минимальный элемент всегда самый левый в дереве, максимальный - самый правый. Значит мы гарантировано знаем куда идти и нам достаточно спуститься к крайнему левому и крайнему правому элементу, если глубина h, то нам нужно h раз спуститься на один элемент, каждая такая операция делается за O(1), значит всего - O(h).

```python
def maxMinMultiplication(data):
    if len(data) < 3:
        return -1
    min_index = 1
    max_index = 2
    i = 0
    while i < len(data):
        min_index = i
        i = 2 * i + 1
    i = 0
    while i < len(data):
        max_index = i
        i = 2 * i + 2
    return data[min_index]*data[max_index]
```

```python
data = [8, 3, 10, 1, 6, 9, 14]
print("Произведение ", maxMinMultiplication(data))

data = [10, 5, 15, 2, 7]
print("Произведение ", maxMinMultiplication(data))
```
